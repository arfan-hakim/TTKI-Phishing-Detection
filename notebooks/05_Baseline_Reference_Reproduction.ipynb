```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json
import platform
import sklearn

from pathlib import Path
```

```python
PROJECT_DIR = Path.cwd().parent

DATA_DIR = PROJECT_DIR / "data"
RESULTS_DIR = PROJECT_DIR / "results"
LOGS_DIR = PROJECT_DIR / "logs"

print("Project Directory:", PROJECT_DIR)
print("Dataset tersedia:", (
    DATA_DIR / "PhiUSIIL_Phishing_URL_Dataset.csv"
).exists())

print("Baseline metrics tersedia:", (
    RESULTS_DIR / "baseline_metrics.csv"
).exists())
```

```python
our_results = pd.read_csv(
    RESULTS_DIR / "baseline_metrics.csv"
)

our_results[
    [
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score",
        "ROC-AUC"
    ]
]
```

```python
reference_results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "Extra Trees"
    ],
    "Reference Accuracy": [
        0.999901,
        1.000000,
        1.000000,
        0.999972
    ],
    "Reference Precision": [
        0.999901,
        1.000000,
        1.000000,
        0.999951
    ],
    "Reference Recall": [
        0.999926,
        1.000000,
        1.000000,
        1.000000
    ],
    "Reference F1": [
        0.999913,
        1.000000,
        1.000000,
        0.999975
    ],
    "Reference ROC-AUC": [
        0.999999,
        1.000000,
        1.000000,
        1.000000
    ]
})

reference_results
```

```python
comparison = our_results.merge(
    reference_results,
    on="Model",
    how="inner"
)

comparison["Accuracy Difference"] = (
    comparison["Accuracy"] -
    comparison["Reference Accuracy"]
)

comparison["F1 Difference"] = (
    comparison["F1-Score"] -
    comparison["Reference F1"]
)

comparison[
    [
        "Model",
        "Accuracy",
        "Reference Accuracy",
        "Accuracy Difference",
        "F1-Score",
        "Reference F1",
        "F1 Difference"
    ]
]
```

```python
reference_actual = pd.read_csv(
    "../results/reference_reproduction_metrics.csv"
)

comparison_actual = reference_actual.merge(
    reference_results,
    on="Model",
    how="inner"
)

comparison_actual["Accuracy Difference"] = (
    comparison_actual["Accuracy"] -
    comparison_actual["Reference Accuracy"]
)

comparison_actual["F1 Difference"] = (
    comparison_actual["F1-Score"] -
    comparison_actual["Reference F1"]
)

display(
    comparison_actual[
        [
            "Model",
            "Accuracy",
            "Reference Accuracy",
            "Accuracy Difference",
            "F1-Score",
            "Reference F1",
            "F1 Difference"
        ]
    ].round(6)
)
```

Reproduksi baseline dilakukan menggunakan dataset PhiUSIIL dan konfigurasi model dari repositori GitHub referensi. Hasil eksperimen menunjukkan bahwa Logistic Regression, Decision Tree, Random Forest, dan Extra Trees memperoleh nilai accuracy serta F1-score yang sama dengan hasil referensi hingga enam angka desimal. Decision Tree dan Random Forest mencapai accuracy dan F1-score sebesar 100%, sedangkan Logistic Regression memperoleh accuracy sebesar 99,9901% dan Extra Trees sebesar 99,9972%.

Hasil tersebut menunjukkan bahwa performa baseline referensi berhasil direproduksi pada tingkat presisi pelaporan yang digunakan. Namun, pencapaian performa yang sangat tinggi belum membuktikan kemampuan generalisasi model terhadap domain yang belum ditemukan selama pelatihan. Oleh karena itu, penelitian selanjutnya akan mengevaluasi pengaruh penggunaan fitur URL-only dan strategi domain-grouped split terhadap performa model deteksi phishing.
